```python
categorical_columns = [
    "order_status",
    "shipping_method",
    "delivery_city",
    "delivery_state",
    "coupon_code",
    "marketing_channel"
]

for col in categorical_columns:
    print("\n" + "=" * 50)
    print("COLUMN:", col)
    print("=" * 50)
    print(df[col].value_counts(dropna=False))
```

```python
id="9v2n0q"
for col in categorical_columns:
    print(
        col,
        "→",
        df[col].notna().sum(),
        "non-missing +",
        df[col].isna().sum(),
        "NaN =",
        len(df)
    )
```

```python
print("Total Order IDs:", df["order_id"].notna().sum())
print("Missing Order IDs:", df["order_id"].isna().sum())
print("Unique Order IDs:", df["order_id"].nunique(dropna=True))
print("Duplicate Order IDs:", df["order_id"].duplicated().sum())
```

```python
invalid_order_ids = ~df["order_id"].str.startswith(
    "ORD",
    na=False
)

print("Invalid Order ID format:", invalid_order_ids.sum())
```

```python
print("Total Customer IDs:", df["customer_id"].notna().sum())
print("Missing Customer IDs:", df["customer_id"].isna().sum())
print("Unique Customer IDs:", df["customer_id"].nunique(dropna=True))
print("Duplicate Customer IDs:", df["customer_id"].duplicated().sum())
```

```python
invalid_customer_ids = ~df["customer_id"].str.startswith(
    "CUST",
    na=False
)

print("Invalid Customer ID format:", invalid_customer_ids.sum())
```

```python
print("Data type before conversion:")
print(df["order_date"].dtype)

print("\nMissing dates:")
print(df["order_date"].isna().sum())

print("\nFirst 10 dates:")
print(df["order_date"].head(10))
```

```python
df["order_date"] = pd.to_datetime(
    df["order_date"],
    errors="coerce"
)

print("Data type after conversion:")
print(df["order_date"].dtype)
```

```python
print("Missing/invalid dates:", df["order_date"].isna().sum())
```

```python
print("Earliest order date:", df["order_date"].min())
print("Latest order date:", df["order_date"].max())
```

```python
print(df["order_date"].head(10))
```

```python
print("Data type before conversion:")
print(df["order_time"].dtype)

print("\nMissing order times:")
print(df["order_time"].isna().sum())

print("\nFirst 10 order times:")
print(df["order_time"].head(10))
```

```python
df["order_time"] = pd.to_datetime(
    df["order_time"],
    format="%H:%M:%S",
    errors="coerce"
).dt.time

print("Order time conversion completed.")
print(df["order_time"].head(10))
```

```python
print(
    "Missing/invalid order times:",
    df["order_time"].isna().sum()
)
```

```python
numeric_columns = [
    "discount_percentage",
    "subtotal",
    "shipping_fee",
    "tax_amount",
    "final_amount"
]

print(df[numeric_columns].dtypes)
```

```python
for col in [
    "subtotal",
    "shipping_fee",
    "tax_amount",
    "final_amount"
]:
    print(
        col,
        "→ Negative values:",
        (df[col] < 0).sum()
    )
```

```python
invalid_discount = (
    (df["discount_percentage"] < 0) |
    (df["discount_percentage"] > 100)
)

print("Invalid discount values:", invalid_discount.sum())
```

```python
id="5x7k2m"
calculated_final = (
    df["subtotal"]
    - (df["subtotal"] * df["discount_percentage"] / 100)
    + df["shipping_fee"]
    + df["tax_amount"]
)

difference = (df["final_amount"] - calculated_final).round(2)

print("Rows matching calculated final amount:", (difference == 0).sum())
print("Rows with differences:", (difference != 0).sum())
print("Maximum difference:", difference.abs().max())
```

```python
print("Missing Campaign IDs:", df["campaign_id"].isna().sum())
print("Unique Campaign IDs:", df["campaign_id"].nunique(dropna=True))

print("\nCampaign ID sample:")
print(df["campaign_id"].value_counts(dropna=False).head(20))
```

```python
campaign_mask = df["campaign_id"].notna()

invalid_campaign_ids = (
    campaign_mask &
    ~df["campaign_id"].str.startswith("CAMP", na=False)
)

print("Invalid Campaign ID format:", invalid_campaign_ids.sum())
```
