```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

df = pd.read_csv("payments.csv")

print("Dataset loaded successfully")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df_original = df.copy()

print("Original backup created successfully")
print("Backup rows:", df_original.shape[0])
print("Backup columns:", df_original.shape[1])
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nDataset information:")
df.info()
```

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": (
        df.isna().sum() / len(df) * 100
    ).round(2)
})

missing_report
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
for col in ["payment_id", "order_id"]:
    print("\n" + "=" * 50)
    print("COLUMN:", col)
    print("=" * 50)

    print("Missing:", df[col].isna().sum())
    print("Unique:", df[col].nunique(dropna=True))
    print("Repeated values:", df[col].duplicated().sum())
```

```python
invalid_payment_ids = ~df["payment_id"].str.startswith("PAY", na=False)
invalid_order_ids = ~df["order_id"].str.startswith("ORD", na=False)

print("Invalid payment_id format:", invalid_payment_ids.sum())
print("Invalid order_id format:", invalid_order_ids.sum())
```

```python
print("Payment date dtype:", df["payment_date"].dtype)
print("Missing dates:", df["payment_date"].isna().sum())

print("\nFirst 10 payment dates:")
print(df["payment_date"].head(10).tolist())
```

```python
df["payment_date"] = pd.to_datetime(
    df["payment_date"],
    errors="coerce"
)

print("Payment date dtype:", df["payment_date"].dtype)
print("Invalid payment dates:", df["payment_date"].isna().sum())

print("Earliest payment date:", df["payment_date"].min())
print("Latest payment date:", df["payment_date"].max())
```

```python
categorical_columns = [
    "payment_method",
    "payment_status"
]

for col in categorical_columns:
    print("\n" + "=" * 50)
    print("COLUMN:", col)
    print("=" * 50)
    print(df[col].value_counts(dropna=False))
```

```python
numeric_columns = [
    "amount_paid",
    "transaction_fee",
    "refund_amount"
]

for col in numeric_columns:
    print("\n" + "=" * 50)
    print("COLUMN:", col)
    print("=" * 50)
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("Zero values:", (df[col] == 0).sum())
    print("Negative values:", (df[col] < 0).sum())
```

```python
zero_amount = df[df["amount_paid"] == 0]

print("Rows with amount_paid = 0:", len(zero_amount))

print("\nPayment status for zero-payment records:")
print(
    zero_amount["payment_status"]
    .value_counts(dropna=False)
)

print("\nPayment method for zero-payment records:")
print(
    zero_amount["payment_method"]
    .value_counts(dropna=False)
)
```

```python
refunded = df[df["payment_status"] == "Refunded"]

print("Refunded payment records:", len(refunded))

print(
    "Refunded records with refund_amount = 0:",
    (refunded["refund_amount"] == 0).sum()
)

print(
    "Refunded records with refund_amount > 0:",
    (refunded["refund_amount"] > 0).sum()
)

print("\nRefund amount summary for Refunded payments:")
print(refunded["refund_amount"].describe())
```

```python
non_refunded = df[df["payment_status"] != "Refunded"]

print(
    "Non-refunded records with refund_amount > 0:",
    (non_refunded["refund_amount"] > 0).sum()
)

print("\nRefund amount by payment status:")
print(
    df.groupby("payment_status")["refund_amount"]
      .agg(["count", "min", "max", "sum"])
)
```

```python
print("Transaction fee summary:")
print(df["transaction_fee"].describe())

print("\nTransaction fee by payment status:")
print(
    df.groupby("payment_status")["transaction_fee"]
      .agg(["count", "min", "max", "mean", "sum"])
)
```

```python
refund_greater_than_paid = df[df["refund_amount"] > df["amount_paid"]]

print("Refund amount greater than amount paid:",
      len(refund_greater_than_paid))

if len(refund_greater_than_paid) > 0:
    print("\nExamples:")
    print(
        refund_greater_than_paid[
            ["payment_id", "order_id", "payment_status",
             "amount_paid", "refund_amount"]
        ].head(10)
    )
```

```python
numeric_cols = ["amount_paid", "transaction_fee", "refund_amount"]

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"\n{col}")
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Outlier count:", len(outliers))
```

```python
print("FINAL INTEGRITY CHECK")
print("=" * 50)

print("Original shape:", df_original.shape)
print("Current shape:", df.shape)

print("Rows removed:", len(df_original) - len(df))
print("Columns removed:", len(df_original.columns) - len(df.columns))

print("\nDuplicate rows:", df.duplicated().sum())
print("Total missing values:", df.isna().sum().sum())

print("\nColumn names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)
```

```python
output_file = "payments_cleaned.csv"

df.to_csv(output_file, index=False)

print("Saved successfully:", output_file)
```
