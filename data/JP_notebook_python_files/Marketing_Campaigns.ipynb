```python
import pandas as pd

df = pd.read_csv("marketing_campaigns.csv")

print("Dataset loaded successfully")
print("Shape:", df.shape)
print("Columns:")
print(df.columns.tolist())
```

```python
df_original = df.copy()

print("Backup created successfully")
print("Original shape:", df_original.shape)
print("Current shape:", df.shape)
```

```python
print("DATASET INFORMATION")
print("=" * 50)

df.info()
```

```python

missing_report = df.isna().sum()

print("MISSING VALUE REPORT")
print("=" * 50)
print(missing_report)

print("\nTotal missing values:", df.isna().sum().sum())
```

```python

duplicate_count = df.duplicated().sum()

print("DUPLICATE RECORD CHECK")
print("=" * 50)
print("Duplicate rows:", duplicate_count)

if duplicate_count > 0:
    print("\nDuplicate records:")
    print(df[df.duplicated(keep=False)])
else:
    print("No duplicate rows found.")
```

```python

print("CAMPAIGN ID VALIDATION")
print("=" * 50)

print("Missing campaign IDs:", df["campaign_id"].isna().sum())
print("Unique campaign IDs:", df["campaign_id"].nunique())
print("Repeated campaign IDs:",
      df["campaign_id"].duplicated().sum())

invalid_campaign_ids = df[
    ~df["campaign_id"].astype(str).str.startswith("CAMP")
]

print("Invalid campaign ID prefix:", len(invalid_campaign_ids))

if len(invalid_campaign_ids) > 0:
    print("\nInvalid campaign IDs:")
    print(invalid_campaign_ids["campaign_id"].tolist())

print("\nSample campaign IDs:")
print(df["campaign_id"].head(10).tolist())
```

```python

categorical_columns = [
    "campaign_name",
    "channel",
    "campaign_type",
    "target_segment"
]

for col in categorical_columns:
    print("\n" + "=" * 50)
    print(f"{col.upper()} — UNIQUE VALUES")
    print("=" * 50)
    print("Unique values:", df[col].nunique())
    print(df[col].value_counts(dropna=False))
```

```python

# Convert dates for validation
start_dates = pd.to_datetime(
    df["campaign_start_date"],
    errors="coerce"
)

end_dates = pd.to_datetime(
    df["campaign_end_date"],
    errors="coerce"
)

print("CAMPAIGN DATE VALIDATION")
print("=" * 50)

print("Invalid start dates:", start_dates.isna().sum())
print("Invalid end dates:", end_dates.isna().sum())

print("Earliest start date:", start_dates.min())
print("Latest start date:", start_dates.max())

print("Earliest end date:", end_dates.min())
print("Latest end date:", end_dates.max())

# Check end date before start date
invalid_date_order = df[
    end_dates < start_dates
]

print("\nCampaigns ending before start:", len(invalid_date_order))

# Campaign duration
campaign_duration = (end_dates - start_dates).dt.days

print("\nCampaign duration statistics:")
print(campaign_duration.describe())
```

```python

print("DISCOUNT PERCENTAGE VALIDATION")
print("=" * 50)

print("Missing:", df["discount_percentage"].isna().sum())
print("Negative values:", (df["discount_percentage"] < 0).sum())
print("Values above 100%:", (df["discount_percentage"] > 100).sum())
print("Zero discount:", (df["discount_percentage"] == 0).sum())

print("\nDiscount statistics:")
print(df["discount_percentage"].describe())

print("\nDiscount value distribution:")
print(df["discount_percentage"].value_counts().sort_index())
```

```python

print("CAMPAIGN PERFORMANCE COUNT VALIDATION")
print("=" * 50)

for col in ["conversions", "impressions", "clicks"]:
    print(f"\n{col}")
    print("Missing:", df[col].isna().sum())
    print("Negative:", (df[col] < 0).sum())
    print("Zero:", (df[col] == 0).sum())
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())

# Business-rule validation
invalid_clicks = df[df["clicks"] > df["impressions"]]
invalid_conversions = df[df["conversions"] > df["clicks"]]

print("\n" + "=" * 50)
print("BUSINESS RULE CHECKS")
print("=" * 50)

print("Clicks greater than impressions:", len(invalid_clicks))
print("Conversions greater than clicks:", len(invalid_conversions))
```

```python

financial_columns = [
    "revenue_generated",
    "campaign_cost"
]

print("REVENUE & CAMPAIGN COST VALIDATION")
print("=" * 50)

for col in financial_columns:
    print(f"\n{col}")
    print("Missing:", df[col].isna().sum())
    print("Negative:", (df[col] < 0).sum())
    print("Zero:", (df[col] == 0).sum())
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())

print("\nFinancial statistics:")
print(df[financial_columns].describe())
```

```python

print("ROI VALIDATION")
print("=" * 50)

print("Missing ROI:", df["roi"].isna().sum())
print("Negative ROI:", (df["roi"] < 0).sum())
print("Zero ROI:", (df["roi"] == 0).sum())

print("\nStored ROI statistics:")
print(df["roi"].describe())

# Calculate ROI for validation only
calculated_roi = (
    (df["revenue_generated"] - df["campaign_cost"])
    / df["campaign_cost"]
) * 100

roi_difference = (
    df["roi"] - calculated_roi
).abs()

print("\nCalculated ROI statistics:")
print(calculated_roi.describe())

print("\nROI consistency check:")
print("Exact/near matches (difference <= 0.01):",
      (roi_difference <= 0.01).sum())

print("Differences greater than 0.01:",
      (roi_difference > 0.01).sum())

print("Maximum ROI difference:",
      roi_difference.max())
```

```python

# Calculate rates for validation only
conversion_rate = (
    df["conversions"] / df["clicks"]
) * 100

click_through_rate = (
    df["clicks"] / df["impressions"]
) * 100

print("CAMPAIGN RATE VALIDATION")
print("=" * 50)

print("Conversion rate:")
print(conversion_rate.describe())

print("\nClick-through rate:")
print(click_through_rate.describe())

print("\nInvalid conversion rates (<0 or >100%):",
      ((conversion_rate < 0) | (conversion_rate > 100)).sum())

print("Invalid click-through rates (<0 or >100%):",
      ((click_through_rate < 0) | (click_through_rate > 100)).sum())
```

```python

numeric_columns = [
    "discount_percentage",
    "conversions",
    "revenue_generated",
    "impressions",
    "clicks",
    "campaign_cost",
    "roi"
]

print("IQR OUTLIER REPORT")
print("=" * 60)

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"\n{col}")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"IQR: {IQR}")
    print(f"Lower bound: {lower_bound}")
    print(f"Upper bound: {upper_bound}")
    print(f"Outliers: {len(outliers)}")
```

```python

print("FINAL INTEGRITY CHECK")
print("=" * 50)

print("Original shape:", df_original.shape)
print("Current shape:", df.shape)

print("Rows removed:", len(df_original) - len(df))
print("Columns removed:", len(df_original.columns) - len(df.columns))

print("Duplicate rows:", df.duplicated().sum())
print("Total missing values:", df.isna().sum().sum())

print("\nMissing values by column:")
print(df.isna().sum())
```
