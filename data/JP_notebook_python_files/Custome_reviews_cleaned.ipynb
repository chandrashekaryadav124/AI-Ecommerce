```python
### inspect duplicates if they exist
df[df.duplicated()]
```

```python
### Check unique values
### This is important for finding inconsistent categories

for column in df.columns:
    print("\n", column)
    print(df[column].unique()[:20])

### checking sentiment
df["review_sentiment"].value_counts(dropna=False)
```

```python
### Validating the rating values
df["rating"].unique()
```

```python
df["rating"].value_counts().sort_index()
```

```python
### If invalid values are found, don't automatically delete them.
### Instead, identify and report them.

df[(df["rating"] < 1) | (df["rating"] > 5)]
```

```python
#### Converting review date to datetime. 
### if now - review_date - object

df["review_date"] = pd.to_datetime(df["review_date"], errors="coerce")
```

```python
### After converting, review_date - datetime64[ns]
df.dtypes
```

```python
### identify invalid dates

df[df["review_date"].isna()]
```

```python
### Validate review dates
### Checking the overall date range.

print("Minimum date:", df["review_date"].min())
print("Maximum date:", df["review_date"].max())
```

```python
df.head()
```

```python
### Checking text columns for unwanted spaces.

df["review_sentiment"].dropna().str.strip().unique()
```

```python
### Standardize them without touching NaN values.

df["review_sentiment"] = df["review_sentiment"].str.strip()
```

```python
### Standardize sentiment values.
### inspect first

df["review_sentiment"].value_counts(dropna=False)
```

```python
### verifying these are the only expected categories: 

valid_sentiments = ["Positive", "Negative", "Neutral"]

invalid_sentiments = df[
    ~df["review_sentiment"].isin(valid_sentiments) &
    df["review_sentiment"].notna()
]

invalid_sentiments
```

```python
### Validate verified_purchase

df["verified_purchase"].value_counts(dropna=False)
```

```python
### Checking for unexpected values

df["verified_purchase"].unique() 
```

```python
### Validate IDs
### Checking whether IDs are missing:

df[["review_id", "order_id", "customer_id", "product_id"]].isnull().sum()
```

```python
### check the number of unique IDs:

print("Unique reviews:", df["review_id"].nunique())
print("Unique orders:", df["order_id"].nunique())
print("Unique customers:", df["customer_id"].nunique())
print("Unique products:", df["product_id"].nunique())
```

```python
### Generate a complete data-quality report:

quality_report = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing_Values": df.isnull().sum().values,
    "Missing_Percentage": (
        df.isnull().sum().values / len(df) * 100
    ),
    "Unique_Values": df.nunique(dropna=True).values
})

quality_report
```
