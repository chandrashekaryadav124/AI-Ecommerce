```python
import pandas as pd

file_path = "products.csv"

df = pd.read_csv(file_path)

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
```

```python
df_original = df.copy()

print("Backup created successfully!")
print("Original shape:", df_original.shape)
print("Current shape:", df.shape)
```

```python
df.info()
```

```python
missing_report = df.isna().sum()

print("Missing values in each column:")
print(missing_report)

print("\nTotal missing values:", missing_report.sum())
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
print("Product ID validation")
print("=" * 30)

print("Missing product IDs:", df["product_id"].isna().sum())
print("Unique product IDs:", df["product_id"].nunique())
print("Repeated product IDs:", df["product_id"].duplicated().sum())
```

```python
invalid_product_ids = (
    ~df["product_id"].str.startswith("PROD")
).sum()

print("Invalid product_id prefix:", invalid_product_ids)

print("\nSample product IDs:")
print(df["product_id"].head(10).tolist())
```

```python
df["product_launch_date"] = pd.to_datetime(
    df["product_launch_date"],
    errors="coerce"
)

print("Data type:", df["product_launch_date"].dtype)

print("Invalid/missing launch dates:",
      df["product_launch_date"].isna().sum())

print("Earliest launch date:",
      df["product_launch_date"].min())

print("Latest launch date:",
      df["product_launch_date"].max())
```

```python
print("Categories:")
print(df["category"].value_counts(dropna=False))

print("\nSubcategories:")
print(df["subcategory"].value_counts(dropna=False))

print("\nBrands:")
print(df["brand"].value_counts(dropna=False))

print("\nProduct Types:")
print(df["product_type"].value_counts(dropna=False))

print("\nDiscount Ranges:")
print(df["discount_range"].value_counts(dropna=False))
```

```python

numeric_cols = [
    "price",
    "cost_price",
    "rating_average",
    "rating_count",
    "stock_quantity",
    "return_rate_baseline"
]

for col in numeric_cols:
    print(f"\n{col}")
    print("-" * 30)
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("Missing:", df[col].isna().sum())
    print("Negative:", (df[col] < 0).sum())
    print("Zero:", (df[col] == 0).sum())
```

```python
cost_greater_than_price = df[df["cost_price"] > df["price"]]

print("Products where cost_price > price:",
      len(cost_greater_than_price))

if len(cost_greater_than_price) > 0:
    print("\nExamples:")
    print(
        cost_greater_than_price[
            [
                "product_id",
                "product_name",
                "price",
                "cost_price",
                "discount_range",
                "product_type"
            ]
        ].head(10)
    )
```

```python

invalid_ratings = df[
    (df["rating_average"] < 0) |
    (df["rating_average"] > 5)
]

print("Invalid rating_average values:", len(invalid_ratings))

print("\nProducts with rating_count = 0:")
print(
    df[df["rating_count"] == 0][
        [
            "product_id",
            "product_name",
            "rating_average",
            "rating_count"
        ]
    ]
)
```

```python

print("Stock quantity validation")
print("=" * 30)
print("Negative stock:", (df["stock_quantity"] < 0).sum())
print("Zero stock:", (df["stock_quantity"] == 0).sum())
print("Minimum stock:", df["stock_quantity"].min())
print("Maximum stock:", df["stock_quantity"].max())

print("\nReturn rate baseline validation")
print("=" * 30)
print("Negative return rates:",
      (df["return_rate_baseline"] < 0).sum())
print("Zero return rates:",
      (df["return_rate_baseline"] == 0).sum())
print("Minimum return rate:",
      df["return_rate_baseline"].min())
print("Maximum return rate:",
      df["return_rate_baseline"].max())
```

```python

import re

# Check discount range format
invalid_discount_format = df[
    ~df["discount_range"].astype(str).str.match(r"^\d+-\d+%$")
]

print("Invalid discount range format:", len(invalid_discount_format))

# Extract minimum and maximum percentages
discount_parts = df["discount_range"].astype(str).str.extract(
    r"^(\d+)-(\d+)%$"
)

discount_min = pd.to_numeric(discount_parts[0], errors="coerce")
discount_max = pd.to_numeric(discount_parts[1], errors="coerce")

# Logical validation
invalid_discount_logic = df[
    (discount_min > discount_max) |
    (discount_min < 0) |
    (discount_max > 100)
]

print("Invalid discount range logic:", len(invalid_discount_logic))

print("\nDiscount ranges:")
print(df["discount_range"].value_counts())
```

```python

# Calculate margin for validation only
margin = df["price"] - df["cost_price"]

print("Negative margin products:", (margin < 0).sum())
print("Zero margin products:", (margin == 0).sum())
print("Positive margin products:", (margin > 0).sum())

print("\nMargin statistics:")
print(margin.describe())

print("\nLowest margins:")
print(
    df.loc[margin.nsmallest(10).index,
           ["product_id", "product_name", "price", "cost_price"]]
    .assign(margin=margin[margin.nsmallest(10).index])
)
```

```python

numeric_columns = [
    "price",
    "cost_price",
    "rating_average",
    "rating_count",
    "stock_quantity",
    "return_rate_baseline"
]

print("IQR Outlier Report")
print("=" * 50)

for col in numeric_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower) |
        (df[col] > upper)
    ]

    print(f"\n{col}")
    print(f"Q1: {Q1}")
    print(f"Q3: {Q3}")
    print(f"IQR: {IQR}")
    print(f"Lower bound: {lower}")
    print(f"Upper bound: {upper}")
    print(f"Outliers: {len(outliers)}")
```

```python

print("FINAL INTEGRITY CHECK")
print("=" * 50)

print("Original shape:", df_original.shape)
print("Current shape:", df.shape)

print("Rows removed:", len(df_original) - len(df))
print("Columns removed:", len(df_original.columns) - len(df.columns))

print("Duplicate rows:", df.duplicated().sum())
print("Total missing values:", df.isna().sum().sum())

print("\nMissing values by column:")
print(df.isna().sum())
```
