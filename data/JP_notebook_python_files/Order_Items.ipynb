```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

df = pd.read_csv("order_items.csv")

print("Dataset loaded successfully")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df_original = df.copy()

print("Original backup created successfully")
print("Backup rows:", df_original.shape[0])
print("Backup columns:", df_original.shape[1])
```

```python
print("Column names:")
print(df.columns.tolist())

print("\nFirst 5 rows:")
display(df.head())

print("\nLast 5 rows:")
display(df.tail())

print("\nDataset information:")
df.info()
```

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": (
        df.isna().sum() / len(df) * 100
    ).round(2)
})

missing_report
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate rows:", duplicate_count)
```

```python
for col in ["order_item_id", "order_id", "product_id"]:
    print("\n" + "=" * 50)
    print("COLUMN:", col)
    print("=" * 50)

    print("Missing:", df[col].isna().sum())
    print("Unique:", df[col].nunique(dropna=True))
    print("Repeated values:", df[col].duplicated().sum())
```

```python
invalid_order_item_ids = ~df["order_item_id"].str.startswith("ITEM", na=False)
invalid_order_ids = ~df["order_id"].str.startswith("ORD", na=False)
invalid_product_ids = ~df["product_id"].str.startswith("PROD", na=False)

print("Invalid order_item_id format:", invalid_order_item_ids.sum())
print("Invalid order_id format:", invalid_order_ids.sum())
print("Invalid product_id format:", invalid_product_ids.sum())
```

```python
print("First 20 order_item_id values:")
print(df["order_item_id"].head(20).tolist())

print("\nSample unique order_item_id values:")
print(df["order_item_id"].drop_duplicates().head(20).tolist())
```

```python
invalid_order_item_ids = ~df["order_item_id"].str.startswith("OI", na=False)

print("Invalid order_item_id format:", invalid_order_item_ids.sum())
```

```python
numeric_columns = [
    "quantity",
    "unit_price",
    "discount_percentage",
    "item_revenue",
    "item_cost",
    "profit"
]

for col in numeric_columns:
    print(
        col,
        "→ Negative values:",
        (df[col] < 0).sum()
    )
    
```

```python
negative_profit = df[df["profit"] < 0]

print("Negative-profit rows:", len(negative_profit))

print("\nNegative-profit summary:")
display(
    negative_profit[
        [
            "quantity",
            "unit_price",
            "discount_percentage",
            "item_revenue",
            "item_cost",
            "profit"
        ]
    ].describe()
)

print("\nFirst 20 negative-profit records:")
display(
    negative_profit[
        [
            "order_item_id",
            "order_id",
            "product_id",
            "quantity",
            "unit_price",
            "discount_percentage",
            "item_revenue",
            "item_cost",
            "profit"
        ]
    ].head(20)
)
```

```python
calculated_profit = (
    df["item_revenue"] - df["item_cost"]
).round(2)

profit_difference = (
    df["profit"] - calculated_profit
).round(2)

print(
    "Rows matching profit formula:",
    (profit_difference == 0).sum()
)

print(
    "Rows with profit differences:",
    (profit_difference != 0).sum()
)

print(
    "Maximum profit difference:",
    profit_difference.abs().max()
)
```

```python
invalid_discount = (
    (df["discount_percentage"] < 0) |
    (df["discount_percentage"] > 100)
)

print("Invalid discount values:", invalid_discount.sum())
```

```python
print("Minimum quantity:", df["quantity"].min())
print("Maximum quantity:", df["quantity"].max())
print("Zero quantity:", (df["quantity"] == 0).sum())
print("Negative quantity:", (df["quantity"] < 0).sum())
```

```python
for col in ["unit_price", "item_revenue", "item_cost"]:
    print("\n" + "=" * 50)
    print("COLUMN:", col)
    print("=" * 50)
    print("Minimum:", df[col].min())
    print("Maximum:", df[col].max())
    print("Zero values:", (df[col] == 0).sum())
    print("Negative values:", (df[col] < 0).sum())
```

```python
calculated_revenue = (
    df["unit_price"]
    * df["quantity"]
    * (1 - df["discount_percentage"] / 100)
).round(2)

revenue_difference = (
    df["item_revenue"] - calculated_revenue
).round(2)

print(
    "Rows matching revenue formula:",
    (revenue_difference == 0).sum()
)

print(
    "Rows with revenue differences:",
    (revenue_difference != 0).sum()
)

print(
    "Maximum revenue difference:",
    revenue_difference.abs().max()
)
```

```python
outlier_columns = [
    "quantity",
    "unit_price",
    "discount_percentage",
    "item_revenue",
    "item_cost",
    "profit"
]

for col in outlier_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = (
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    )

    print("\n" + "=" * 50)
    print("COLUMN:", col)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower Bound:", lower_bound)
    print("Upper Bound:", upper_bound)
    print("Potential Outliers:", outliers.sum())
```

```python
print("Original shape:", df_original.shape)
print("Current shape:", df.shape)

print("\nRows removed:",
      df_original.shape[0] - df.shape[0])

print("Columns removed:",
      df_original.shape[1] - df.shape[1])

print("\nCurrent duplicate rows:",
      df.duplicated().sum())

print("Current total missing values:",
      df.isna().sum().sum())
```

```python
df.to_csv("order_items_cleaned.csv", index=False)

print("order_items_cleaned.csv created successfully")
```
