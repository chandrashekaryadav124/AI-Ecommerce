```python
df["customer_signup_date"] = pd.to_datetime(
    df["customer_signup_date"],
    errors="coerce"
)

df["last_order_date"] = pd.to_datetime(
    df["last_order_date"],
    errors="coerce"
)
```

```python
#Last order before signup
invalid_dates = df[
    df["last_order_date"] < df["customer_signup_date"]
]

invalid_dates
```

```python
today = pd.Timestamp.today()

future_signup = df[
    df["customer_signup_date"] > today
]

future_signup
```

```python
df["calculated_aov"] = np.where(
    df["total_orders"] > 0,
    df["total_spend"] / df["total_orders"],
    np.nan
)

df["calculated_aov"] 
```

```python
df[[
    "total_orders",
    "total_spend",
    "average_order_value",
    "calculated_aov"
]].head()
```

```python
df["aov_difference"] = (
    df["average_order_value"] -
    df["calculated_aov"]
).abs()
```

```python
df["aov_difference"]
```

```python
# Outlier detection : 
Q1 = df["total_spend"].quantile(0.25)
Q3 = df["total_spend"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["total_spend"] < lower_bound) |
    (df["total_spend"] > upper_bound)
]

print("Total spend outliers:", len(outliers))
```

# Report :

```python
validation_report = pd.DataFrame({
    "check": [
        "Duplicate Rows",
        "Missing Values",
        "Duplicate Customer IDs",
        "Invalid Age",
        "Negative Total Orders",
        "Negative Total Spend",
        "Invalid Signup Dates",
        "Future Signup Dates"
    ],
    "result": [
        df.duplicated().sum(),
        df.isnull().sum().sum(),
        df["customer_id"].duplicated().sum(),
        ((df["age"] < 0) | (df["age"] > 120)).sum(),
        (df["total_orders"] < 0).sum(),
        (df["total_spend"] < 0).sum(),
        df["customer_signup_date"].isna().sum(),
        (df["customer_signup_date"] > pd.Timestamp.today()).sum()
    ]
})

validation_report
```

```python
validation_report["status"] = np.where(
    validation_report["result"] == 0,
    "PASS",
    "REVIEW"
)

validation_report
```

```python
df.to_csv(
    "customers_cleaned.csv",
    index=False
)

```
